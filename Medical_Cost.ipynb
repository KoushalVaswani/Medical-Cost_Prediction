```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
data['sleep_hours'] = scaler.fit_transform(data[['sleep_hours']])
data['bmi'] = scaler.fit_transform(data[['bmi']])
data['daily_steps'] = scaler.fit_transform(data[['daily_steps']])
data['medication_count'] = scaler.fit_transform(data[['medication_count']])
data['hospital_admissions'] = scaler.fit_transform(data[['hospital_admissions']])
data['doctor_visits_per_year'] = scaler.fit_transform(data[['doctor_visits_per_year']])
data['age'] = scaler.fit_transform(data[['age']])
```

```python
data['physical_activity_level_Low'] = data['physical_activity_level_Low'].astype(int)
data['physical_activity_level_Medium'] = data['physical_activity_level_Medium'].astype(int)
display(data.head())
```

```python
data['city_type_Semi-Urban'] = data['city_type_Semi-Urban'].astype(int)
data['city_type_Urban'] = data['city_type_Urban'].astype(int)
display(data.head())
data['previous_year_cost'] = scaler.fit_transform(data[['previous_year_cost']])
data
```

```python
from sklearn.model_selection import train_test_split
x = data.drop('annual_medical_cost' , axis = 1)
y = data['annual_medical_cost']
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2 , random_state = 42)

```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error , r2_score , mean_absolute_error , accuracy_score
model = LinearRegression()
model.fit(x_train , y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
mse = mean_squared_error(y_test , y_pred)
r2 = r2_score(y_test , y_pred)
adjusted_r2 = 1 - (1 - r2) * (len(y_test) - 1) / (len(y_test) - x_test.shape[1] - 1)
r2
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error
import numpy as np

mse_rf = mean_squared_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mse_rf)
mape_rf = mean_absolute_percentage_error(y_test, y_pred_rf)

print(f"MSE: {mse_rf:.2f}")
print(f"RMSE: {rmse_rf:.2f}")
print(f"MAPE: {mape_rf * 100:.2f}%")
```
